```python
import os
import csv
import pandas as pd
import matplotlib.pyplot as plt

path = "C:\\Gen AI Course Details\\My Project"
myfile = "Sports_Access.log"
```

```python
Ip_Address = []
Error_Status = []

with open(os.path.join(path,myfile)) as n:
    for each_word in n:
        mylist = each_word.split(' ')
        Ip_Address.append(mylist[0])
        Error_Status.append(mylist[8])
```

```python
def frequency(mylist):
    result = {}
    for each in mylist:
        if each not in result:
            result[each] = 1
        else:
            result[each] += 1
    return result 
```

```python
def Ip_to_csv(data):
    with open(os.path.join(path,'Ip_Access.csv'),'w',newline = '') as n:
        writer = csv.writer(n)
        writer.writerow(['S.No','Ip_Address','Frequency'])
        Sno = 1
        for each in data:
            writer.writerow([Sno,each,data[each]])
            Sno += 1
```

```python
def Status_to_csv(data):
    with open(os.path.join(path,'E.Status_Access.csv'),'w',newline = '') as n:
        writer = csv.writer(n)
        writer.writerow(['S.No','Error_Status','Frequency'])
        Sno = 1
        for each in data:
            writer.writerow([Sno,each,data[each]])
            Sno += 1
```

```python
obj = frequency(Ip_Address)
obj1 = frequency(Error_Status)   
        
Ip_to_csv(obj)
Status_to_csv(obj1)

print("Unique IPs and their counts:")
print(obj)
print("\nHTTP Status code counts:")
print(obj1)

print('CSV file created sucessfully')
```
